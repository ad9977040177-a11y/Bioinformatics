```python
def count_dna_bases(sequence):
    """
    Counts the occurrences of A, T, G, C in a DNA sequence.
    Args:
        sequence (str): The DNA sequence string.
    Returns:
        dict: A dictionary with counts for 'A', 'T', 'G', 'C'.
    """
    counts = {'A': 0, 'T': 0, 'G': 0, 'C': 0}
    for base in sequence.upper(): # Convert to uppercase to handle mixed case input
        if base in counts:
            counts[base] += 1
    return counts

# Example usage:
dna_sequence = "ATGCGTACGTACGTAGCTAGCTAGC"
base_counts = count_dna_bases(dna_sequence)
print(f"DNA Sequence: {dna_sequence}")
print(f"Base Counts: {base_counts}")
```
